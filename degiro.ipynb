```python
activa = 209.97
passiva = 91.16	
schulden = 64.54	
eigen_vermogen = activa - passiva
totaal_vermogen = activa
nettowinst = -17.85
verwachte_winst = -15.159
uitstaande_aandelen = 240.446
actuele_beurskoers = 0.747
dividend_rendement = 0
tienjaars_staatsobligatie = 2.7710 
boekwaarde_per_aandeel = 0.45
```

```python
checkboxes = 0
```

Voor de solvabiliteit is tussen 25%-50% goed.

```python
solvabiliteit = (eigen_vermogen / totaal_vermogen) * 100
if 50 >= solvabiliteit >= 25:
    checkboxes += 1
else:
    print(f'Not check: {solvabiliteit}')
```

Voor de rentabiliteit geldt dat wanneer het groter is dan het rendement op de 10jaars staatsobligaties+10%, het goed is.


```python
rentabiliteit = (uitstaande_aandelen / totaal_vermogen) * 100
print(rentabiliteit)
if rentabiliteit > tienjaars_staatsobligatie + 10:
    checkboxes += 1
else:
    print(f"Not check:{rentabiliteit} < {tienjaars_staatsobligatie}")
```

```python
wpa = nettowinst / uitstaande_aandelen
print(wpa)
```

Deze ratio wordt vaak gebruikt om bedrijven binnen dezelfde sector te vergelijken, omdat waarderingen tussen sectoren sterk kunnen verschillen.

Een hogere K/W-verhouding kan duiden op hoge groeiverwachtingen of op een overwaardering, terwijl een lagere verhouding kan wijzen op onderwaardering of zwakke groeivooruitzichten. Echter, de interpretatie moet steeds gebeuren in samenhang met andere factoren, zoals winststabiliteit en sectorale dynamiek.

Wanneer de koers-winst verhouding 10 of lager is, dan is het een goed moment om te kopen. Wanneer het getal tussen de 10-20, raden ze het aan om het aandeel alleen houden.


```python
kw = actuele_beurskoers / wpa
if kw < 10:
    checkboxes += 1
    print(f'Koers-winstverhouding:{kw}')
elif 20 >= kw >= 10:
    checkboxes += 0.5
    print(kw)
else:
    print(f'Not check: {kw}')  
```

Deze verhouding biedt een genuanceerder beeld omdat ze rekening houdt met groeipotentieel. Een PEG van ongeveer 1 wordt vaak als “neutraal” beschouwd: de waardering staat dan in verhouding tot de verwachte groei. Echter, interpretaties zoals “ondergewaardeerd” of “overgewaardeerd” bij respectievelijk PEG < 1 of > 1 dienen voorzichtig te gebeuren en altijd binnen context geplaatst te worden.

```python
peg = kw / verwachte_winst
if peg < 1:
    checkboxes += 1
    print(f'Het aandeel is ondergewaardeerd: {peg}')
else:
    print(f'Het aandeel is overgewaardeerd: {peg}')
```

The graham number geeft de maximale koers aan voor wat je zou moeten betalen voor het aandeel.


```python
import math
graham_number = math.sqrt(22.5*wpa*boekwaarde_per_aandeel)
if actuele_beurskoers - graham_number > 0: 
    verschil = actuele_beurskoers - graham_number
    print(f'Het aandeel is {verschil} te duur')
else:
    verschil = graham_number - actuele_beurskoers
    print(f'Het aandeel ligt met €{verschil} onder de maximale koers')
print(graham_number)
```
